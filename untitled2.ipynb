2. Physical parameters:

* $g  = 9.81$       (gravitational acceleration [m/s²])
* $l  = 1.0$         (pendulum length [m])
* $w0 = np.sqrt(g/l)$    (natural frequency [rad/s])
* $wd = 2 * w0$         (driving frequency [rad/s])

* t_max = 20.0        # total simulation time [s]
* dt = 0.001          # time step [s]
* t_eval = np.arange(0, t_max, dt)

      print(f"Natural frequency $\omega_0$ = {w0:.3f} rad/s, drive frequency $\omega_d$ = {wd:.3f} rad/s")

3.  Define pivot motion:

* def pivot_position(t):

   (Return pivot position (xp, yp) at time t.)

   xp = -l * np.sin(wd * t)

   yp =  l * (np.cos(wd * t) - 1)

      return xp, yp

* def pivot_acceleration(t):
   
   (Return pivot acceleration (xpp, ypp) at time t.)

   xpp =  l * wd**2 * np.sin(wd * t)

   ypp = -l * wd**2 * np.cos(wd * t)

      return xpp, ypp

4. Equation of motion

* Derived from Lagrangian mechanics:

     $l\omega''+ g sinθ = −x''p cosθ − y''p sinθ$
    
     so that $\omega''= −(g/l) sinθ − (x''p/l) cosθ − (y''p/l) sinθ$

* def pendulum_rhs(t, y):

     (Return [$\omega' ;\omega''$] for the driven pendulum.)

     $\theta, \theta' = y$

     xpp, ypp = pivot_acceleration(t)

     $\theta'' = -(g/l)*np.sin(\theta) - (xpp/l)*np.cos(\theta) - (ypp/l)*np.sin(\theta)$

      return [theta', theta''$]

5. Initial and integration

* $\theta_0 = 1e-6$       small displacement [rad]
* $\theta'_0 = 0.0$    initial angular velocity [rad/s]
* $y_0 = [\theta0, \theta'_0]$

      sol = solve_ivp(pendulum_rhs, [0, t_max], y0, t_eval=t_eval, rtol=1e-8, atol=1e-10)

* theta = sol.y[0]
* theta_dot = sol.y[1]
* time = sol.t

      print("Integration complete. Sample θ(t):", theta[:5])

* x_lab = xp + x_rel
