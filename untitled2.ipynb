```python
# Imports

import numpy as np
import matplotlib.pyplot as plt
from scipy.integrate import solve_ivp
from IPython.display import HTML
from matplotlib.animation import FuncAnimation

# Configure inline graphics
plt.rcParams["figure.figsize"] = (6, 4)
plt.rcParams["font.size"] = 13
```

```python
#Define physical parameters

# constants
g  = 9.81       # gravitational acceleration [m/s²]
l  = 1.0        # pendulum length [m]
w0 = np.sqrt(g/l)  # natural angular frequency [rad/s]

# Driving frequency (can change later for analysis)
wd = 2 * w0

# Simulation time parameters
t_max = 10.0       # total simulation time [s]
dt    = 0.001       # time step [s]
t_eval = np.arange(0, t_max, dt)

print(f"Natural frequency w0 = {w0:.3f} rad/s, drive frequency wd = {wd:.3f} rad/s")
```

```python
# Define pivot motion and accelerations

def pivot_position(t):
    """Return pivot (xp, yp) at time t."""
    xp = -l * np.sin(wd * t)
    yp =  l * (np.cos(wd * t) - 1)
    return xp, yp

def pivot_acceleration(t):
    """Return pivot accelerations (xpp, ypp) at time t."""
    xpp =  l * wd**2 * np.sin(wd * t)
    ypp = -l * wd**2 * np.cos(wd * t)
    return xpp, ypp
```

```python
#Equation of motion

# Derived from Lagrangian mechanics:
#    l θ'' + g sinθ = −x''p cosθ − y''p sinθ
# => θ''= −(g/l) sinθ − (x''p/l) cosθ − (y''p/l) sinθ

def pendulum_rhs(t, y):
    """Right-hand side of pendulum ODE.
    y[0] = θ, y[1] = θ'
    Returns derivatives [θ̇, θ'].
    """
    theta, theta_dot = y
    xpp, ypp = pivot_acceleration(t)
    theta_ddot = -(g/l) * np.sin(theta) - (xpp/l) * np.cos(theta) - (ypp/l) * np.sin(theta)
    return [theta_dot, theta_ddot]
```

```python
# Initial conditions and intergration

theta0 = 1e-6     # small initial displacement [rad]
theta_dot0 = 0.0  # initial angular velocity [rad/s]
y0 = [theta0, theta_dot0]

sol = solve_ivp(pendulum_rhs, [0, t_max], y0, t_eval=t_eval, rtol=1e-8, atol=1e-10)
theta = sol.y[0]
theta_dot = sol.y[1]
time = sol.t

print("Integration complete. Sample θ(t):", theta[:5])
```

```python
# Convert to cartesian coordinates

xp, yp = pivot_position(time)
x_rel =  l * np.sin(theta)
y_rel = -l * np.cos(theta)

x_lab = xp + x_rel
y_lab = yp + y_rel
```

```python
# Plot angle vs time

plt.figure()
plt.plot(time, theta, lw=1)
plt.title("Angular Displacement vs Time")
plt.xlabel("Time [s]")
plt.ylabel("θ [rad]")
plt.grid(True)
plt.show()
```

```python
# Animation function

def create_pendulum_animation(time, x_lab, y_lab, xp, yp):
    """
    Create an animated pendulum motion plot in the lab frame.
    """
    fig, ax = plt.subplots(figsize=(6, 6))
    ax.set_xlim(-2 * l, 2 * l)
    ax.set_ylim(-2 * l, 2 * l)
    ax.set_aspect('equal')
    ax.set_title('Circularly Driven Pendulum', fontsize=14)
    ax.set_xlabel('x (m)')
    ax.set_ylabel('y (m)')

    # Static elements
    pivot_dot, = ax.plot([], [], 'ko', markersize=6, label='Pivot')
    line, = ax.plot([], [], 'b-', lw=2)
    bob, = ax.plot([], [], 'ro', markersize=8)
    traj, = ax.plot([], [], 'r--', lw=1, alpha=0.6)

    # Initialize
    def init():
        pivot_dot.set_data([], [])
        line.set_data([], [])
        bob.set_data([], [])
        traj.set_data([], [])
        return line, bob, pivot_dot, traj

    # Update each frame
    def update(frame):
        # Each of these must be sequences (not scalars)
        pivot_dot.set_data([xp[frame]], [yp[frame]])
        line.set_data([xp[frame], x_lab[frame]], [yp[frame], y_lab[frame]])
        bob.set_data([x_lab[frame]], [y_lab[frame]])
        traj.set_data(x_lab[:frame], y_lab[:frame])
        return line, bob, pivot_dot, traj

    ani = FuncAnimation(fig, update, frames=len(time), init_func=init,
                        blit=True, interval=10, repeat=False)
    plt.close(fig)
    return HTML(ani.to_html5_video())
```

```python
# Run animation

animation_html = create_pendulum_animation(time, x_lab, y_lab, xp, yp)
display(animation_html)
```

```python
# Plot theta(t) vs t

plt.figure(figsize=(8, 4))
plt.plot(time, theta, color='darkblue')
plt.title("Pendulum Angular Displacement vs. Time")
plt.xlabel("Time (s)")
plt.ylabel("θ (radians)")
plt.grid(True)
plt.show()
```

Specifically: If the pendulum bob starts from rest $(\theta_0 = 0, \theta_0' = 0)$ , will it remain nearly stationary? (Note that in the lab frame, the pendulum bob's initial position is (0, - l).)

No, $\theta(t) = 0$ (bob exactly hanging still) is not a true solution for the given pivot motion (except for the trivial case of zero driving). If you start at $(\theta_0 = 0, \theta_0' = 0)$ the bob will not remain forever at rest: the pivot acceleration directly forces the pendulum.

Under what drive frequencies or amplitudes might this equilibrium remain stable or become unstable?

Whether the resulting motion stays small and bounded or grows exponentially depends on the drive parameters. The linearized equation has a parametric instability (exponential growth of small deviations) which appears in frequency bands. The dominant band is near $\omega_d = 2\omega_0, \omega_0 = \sqrt{g/l}$, and other tongues occur near $\omega_d = \frac{2\omega_0}{n}$ for integer
n. The width and severity of the instability depend on the drive amplitude (here the pivot radius) and any damping.
