Goal:
* Compute the motion of a rigid pendulum (length l) whose pivot moves on a circle of radius l around the pendulum’s initial equilibrium point.
* Start from $\theta(0) = 0, \theta'(0) = 0$ (bob hanging straight down relative to the instantaneous pivot at t = 0).
* Answer two linked questions:
  * Does the bob remain “nearly stationary” (in the lab frame) under this drive?
  * For which drive parameters (frequency / amplitude) does the response stay small vs grow large (instability / resonance-like behavior)?



Psuedo code:

1. Define physical constants and configuration
   * set g, pendulum length l
   * compute natural frequency omega0 = sqrt(g/l)
   * choose drive parameters (omega_d; optionally drive radius A)

2. Define prescribed pivot motion (given in the prompt)
   * function pivot_xy(t): return x_p(t), y_p(t)
   * function pivot_xpp_ypp(t): return x¨_p(t), y¨_p(t)

3. Define geometry helpers (convert between frames)
   * function pendulum_rel(theta): return (x_rel, y_rel) = (l sinθ, -l cosθ)
   * function lab_xy(x_p, y_p, x_rel, y_rel): return (x_lab, y_lab)

4. Define equation of motion for θ(t) (RHS for ODE integrator)
   * state vector y = [θ, θdot]
   * compute x¨_p(t), y¨_p(t)
   * compute θ¨ using:
          θ¨ = -(g/l) sinθ - (sinθ * y¨_p + cosθ * x¨_p)/l
   * return [θdot, θ¨]  (this is f(t, y))

5. Numerical integration (base case: start from rest)
   * set initial condition y0 = [0, 0]
   * choose time interval [t_min, t_max] and evaluation grid t_eval
   * call solve_ivp with:
         - dense_output = True (for smooth animation sampling)
         - tight tolerances (rtol/atol) to avoid drift when driving is fast
   * store solution θ(t), θdot(t)

6. Lab-frame reconstruction (for plots + animation)
   * pick t_plot (dense grid)
   * evaluate θ(t_plot) via dense output
   * compute pivot positions (x_p, y_p) on t_plot
   * compute bob position relative to pivot (x_rel, y_rel) from θ(t_plot)
   * compute lab positions (x_lab, y_lab) = lab_xy(...)

7. Visualization
   * plot θ(t) vs t (diagnostic: does it remain small?)
   * build animation using create_pendulum_animation(t_plot, x_p, y_p, x_lab, y_lab)

8. Stability / resonance diagnosis (frequency sweep)
   * define a set of ratios r = omega_d / omega0 (e.g., 0.2 → 3.0)
   * for each r:
         - set omega_d = r * omega0
         - integrate from y0=[0,0] over a fixed horizon T_sweep
         - compute metric M(r) = max(|θ(t)|) over that run
   * plot M(r) vs r (peaks identify dangerous bands)

9. Integration (what to write in the report)
   * explain why “stationary” in lab frame is not a true equilibrium:
         pivot has nonzero acceleration → inertial forcing appears in θ-equation
   * connect large peaks in M(r) to:
         - direct resonance: ω_d ≈ ω0
         - parametric resonance window: ω_d ≈ 2 ω0 (and nearby bands)
   * conclude: for drive frequencies far from these bands (especially ω_d >> ω0),
         motion stays comparatively small (practically “stable” over the chosen horizon).




```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.integrate import solve_ivp
from matplotlib.animation import FuncAnimation
from IPython.display import HTML

np.set_printoptions(precision=6, suppress=True)
```

```python
# physical parameters
g = 9.81           # m/s^2
l = 1.0            # m (also the pivot circle radius, per statement)
omega0 = np.sqrt(g/l)   # natural small-angle frequency (pendulum about downward equilibrium)

# drive parameters (choose one set; or sweep over them later)
omega_d = 1.2 * omega0   # drive frequency (rad/s)

# time window for simulation
t_min, t_max = 0.0, 30.0   # seconds
t_eval = np.linspace(t_min, t_max, 4000)

def pivot_xy(t, omega_d=omega_d, A=0.1*l):
    x_p = -A * np.sin(omega_d * t)
    y_p =  A * (1 - np.cos(omega_d * t))
    return x_p, y_p

def pivot_xpp_ypp(t, omega_d=omega_d, A=0.1*l):
    xpp =  A * (omega_d**2) * np.sin(omega_d * t)
    ypp =  A * (omega_d**2) * np.cos(omega_d * t)
    return xpp, ypp

def pendulum_xy_rel(theta, l=l):
    # coordinates of bob relative to pivot (θ measured from downward vertical)
    x_rel = l * np.sin(theta)
    y_rel = -l * np.cos(theta)
    return x_rel, y_rel

def lab_xy(x_p, y_p, x_rel, y_rel):
    return x_p + x_rel, y_p + y_rel
```

```python
def pendulum_rhs(t, y, g=g, l=l, omega_d=omega_d, A=0.1*l):
    theta, theta_dot = y
    xpp, ypp = pivot_xpp_ypp(t, omega_d=omega_d, A=A)
    theta_ddot = -(g/l)*np.sin(theta) - (np.sin(theta)*ypp + np.cos(theta)*xpp)/l
    return [theta_dot, theta_ddot]

# initial state: at rest from the downward configuration (relative to the pivot)
y0 = [0.0, 0.0]

sol = solve_ivp(
    fun=lambda t, y: pendulum_rhs(t, y, g=g, l=l, omega_d=omega_d),
    t_span=(t_min, t_max),
    y0=y0,
    t_eval=t_eval,
    dense_output=True,
    rtol=1e-9,
    atol=1e-12,
    method='DOP853'
)

print('integrator success:', sol.success)
print('omega0 =', omega0, 'rad/s  |  omega_d =', omega_d, 'rad/s')
```

```python
t = sol.t
theta = sol.y[0]
theta_dot = sol.y[1]

fig, ax = plt.subplots(figsize=(7,3.2))
ax.plot(t, theta, lw=2)
ax.set_xlabel('t [s]')
ax.set_ylabel(r'$\theta(t)$ [rad]')
ax.axhline(0, color='k', lw=1)
ax.set_title('Pendulum angle (initially at rest, $\theta(0)=\dot\theta(0)=0$)')
plt.show()

print('max |theta| over interval =', np.max(np.abs(np.sin(theta))))
```

```python
def create_pendulum_animation(t, x_pivot, y_pivot, x_lab, y_lab, l=1, show_traj=False, speed_factor=1.0):
    """Creates an animation of the pivot-driven pendulum."""
    fig, ax = plt.subplots(figsize=(4, 4))
    ax.set_xlim(np.min(x_pivot) - l, np.max(x_pivot) + l)
    ax.set_ylim(np.min(y_pivot) - l, np.max(y_pivot) + l)
    ax.set_aspect('equal')
    ax.set_xlabel('x')
    ax.set_ylabel('y')
    pivot, = ax.plot([], [], 'o', color='black', markersize=8, label='Pivot')
    pendulum_arm, = ax.plot([], [], '-', color='black', lw=2, label='Pendulum Arm')
    pendulum_bob, = ax.plot([], [], 'o', color='red', markersize=12, label='Pendulum Bob')
    trajectory = None
    if show_traj:
        trajectory, = ax.plot([], [], '-', color='gray', lw=1, alpha=0.5, label='Trajectory')
    def animate(i):
        pivot.set_data([x_pivot[i]], [y_pivot[i]])
        pendulum_arm.set_data([x_pivot[i], x_lab[i]], [y_pivot[i], y_lab[i]])
        pendulum_bob.set_data([x_lab[i]], [y_lab[i]])
        artists = [pivot, pendulum_arm, pendulum_bob]
        if show_traj and trajectory:
            trajectory.set_data(x_lab[:i+1], y_lab[:i+1])
            artists.append(trajectory)
        return artists
    average_time_step = np.mean(np.diff(t))
    interval = average_time_step * 1000 * speed_factor
    anim = FuncAnimation(fig, animate, frames=len(t), interval=interval, blit=True)
    plt.close(fig)
    return anim

# build coordinates for animation (using a dense time grid)
n_points = 600
t_plot = np.linspace(t_min, t_max, n_points)
theta_plot = sol.sol(t_plot)[0]
x_pivot, y_pivot = pivot_xy(t_plot, omega_d=omega_d)
x_rel, y_rel = pendulum_xy_rel(theta_plot, l=l)
x_lab, y_lab = lab_xy(x_pivot, y_pivot, x_rel, y_rel)

pendulum_animation = create_pendulum_animation(
    t_plot, x_pivot, y_pivot, x_lab, y_lab,
    l=l, show_traj=True, speed_factor=5.0
)
HTML(pendulum_animation.to_html5_video())
```

```python
def max_excursion_wrapped(theta_array):
    theta_wrapped = (theta_array + np.pi) % (2*np.pi) - np.pi
    return np.max(np.abs(theta_wrapped))

def run_case(omega_d_ratio, t_max=40.0, A=0.1*l):
    omega_d = omega_d_ratio * omega0
    sol = solve_ivp(
        fun=lambda t, y: pendulum_rhs(t, y, g=g, l=l, omega_d=omega_d, A=A),
        t_span=(0.0, t_max),
        y0=[0.0, 0.0],
        t_eval=np.linspace(0.0, t_max, 4000),
        rtol=1e-9, atol=1e-12, method='DOP853'
    )
    return max_excursion_wrapped(sol.y[0])

ratios = np.linspace(0.2, 3.0, 34)
max_theta = np.array([run_case(r) for r in ratios])

fig, ax = plt.subplots(figsize=(7,3.2))
ax.plot(ratios, max_theta, marker='o')
ax.set_xlabel(r'$\omega_d / \omega_0$')
ax.set_ylabel(r'$\max\,|\theta|$ [rad]$')
ax.set_title('Frequency sweep (initially at rest): larger peaks = stronger response')
ax.grid(True, alpha=0.3)
plt.show()
```

#End discussion:

For a rigid pendulum with its pivot moving in a circular path about the initial equilibrium, the bob starting from rest at the vertical downward position will **not remain perfectly stationary**. The circular motion of the pivot continuously applies a time-dependent forcing to the bob, even if the initial angular displacement and velocity are zero.

The stability of the downward equilibrium depends critically on the driving frequency $\omega_d$ and the effective amplitude (here set by the pivot radius l):
* **Stable regime**: When the driving frequency is small compared to the natural frequency of the pendulum $\omega_d < \sqrt{g/l}$ or is far from parametric resonance conditions, the bob remains near the vertical position, executing small oscillations.
* **Unstable regime**: If the driving frequency approaches parametric resonance values $(\omega_d = 2\sqrt{g/l}, 4\sqrt{g/l},...)$ or the pivot radius is large, the equilibrium becomes unstable and the pendulum exhibits increasingly large oscillations.

In essence, the vertical downward position is conditionally stable: the bob can remain nearly stationary only when the pivot motion avoids resonance conditions; otherwise, the pivot’s circular driving rapidly destabilizes the pendulum.
